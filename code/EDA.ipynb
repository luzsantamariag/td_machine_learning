1. **Skewness (asimetría)**

La skewness indica si los datos están distribuidos de manera simétrica o si tienen una cola más larga hacia un lado.

- Skewness = 0: distribución aproximadamente simétrica.
- Skewness > 0: asimetría positiva; la cola se extiende hacia la derecha.
- Skewness < 0: asimetría negativa; la cola se extiende hacia la izquierda.

Un ejemplo típico sería ingresos: muchas personas pueden tener ingresos relativamente bajos/medios, mientras que unas pocas tienen ingresos extremadamente altos. Esos valores altos generan una cola hacia la derecha.

2. **Kurtosis (curtosis)**

La kurtosis mide principalmente qué tan pesadas son las colas de una distribución, es decir, qué tan propensos son los datos a presentar valores extremos (outliers).

Una forma común de expresarla es mediante excess kurtosis:

- Kurtosis = 0: similar a una distribución normal.
- Kurtosis > 0: colas más pesadas → mayor presencia de valores extremos.
- Kurtosis < 0: colas más ligeras → menos valores extremos.


```python
# Skewness (asimetría) y kurtosis (curtosis)
print("Skewness: {:0.3f}".format(housing['SalePrice'].skew()))
print("Kurtosis: {:0.3f}".format(housing['SalePrice'].kurt()))
```

Esos valores indican una distribución bastante asimétrica y con colas pesadas.

**Skewness = 1.883**
Es positiva y relativamente alta.

Esto significa que tus datos tienen una cola larga hacia la derecha:

La mayoría de los valores probablemente están concentrados en la parte baja/media.
Hay algunos valores mucho más grandes que el resto.
Es probable que existan valores extremos altos.
La media puede estar siendo elevada por esos valores extremos.

**Kurtosis = 6.536**
Aquí hay una pequeña consideración: hay dos convenciones.

Si 6.536 es excess kurtosis, es bastante alta. Significa que la distribución tiene colas pesadas y una mayor presencia de valores extremos que una distribución normal.

Si 6.536 es kurtosis "normal" (Pearson), sigue siendo alta, porque una distribución normal tiene kurtosis 3. En ese caso, el excess kurtosis sería aproximadamente:

6.536 − 3 = 3.536

En ambos casos hay evidencia de valores extremos/outliers y la distribución no se parece mucho a una normal.

En otras palabras, hay muchas casas con precios relativamente bajos/medios y algunas casas con precios muy altos que generan una cola hacia la derecha. Enel dataset Ames Housing, SalePrice no sigue una distribución normal y presenta una clara concentración de precios hacia valores bajos/medios con una cola de precios muy altos.

Esto es importante porque, en modelos estadísticos que asumen normalidad, suele ser útil transformar SalePrice, dependiendo del modelo que se vaya a utilizar y de lo que se requiere explorar.

```python
# Identificar qué categorías de cada variable categórica 
#  aparecen al menos 30 veces. 
#  Es una técnica útil para saber qué niveles tienen 
#  suficiente cantidad de observaciones.
def identify_cat_above30(series):
    counts = series.value_counts()          # Cuenta cuántas veces aparece cada valor.
    return list(counts[counts>=30].index)   # Obtiene los nombres de las categorías
```

```python
# Selecciona únicamente las variables categóricas del dataset
levels_to_keep = housing[categorical_vars].apply(identify_cat_above30, axis=0)
levels_to_keep
```

Por ejemplo, [RL, RM, FV] son las categorías de MSZoning que aparecen 30 o más veces.

```python
# Reemplaza el DataFrame original por el DataFrame filtrado
for var in categorical_vars:
    housing = housing.loc[housing[var].isin(levels_to_keep[var])]
```

## Relación del Precio de Venta con Variables Categóricas

Aquí nos interesa la relación entre las variables categóricas de nuestro conjunto de datos y el Precio de Venta de la vivienda. El gráfico estándar para examinar la relación entre una variable numérica y una categórica es el diagrama de caja. Un diagrama de caja es una forma práctica de representar gráficamente grupos de datos numéricos mediante sus cuartiles. Los diagramas de caja también pueden incluir líneas verticales que se extienden desde las cajas (bigotes), indicando la variabilidad fuera de los cuartiles superior e inferior; de ahí los términos diagrama de caja y bigotes. Los valores atípicos pueden representarse como puntos individuales. Los diagramas de caja son no paramétricos: muestran la variación en muestras de una población estadística sin asumir ninguna distribución estadística subyacente.

```python
sns.boxplot(x='CentralAir', y='SalePrice', data=housing)
```

El diagrama de cajas permite comparar la distribución de SalePrice entre las viviendas que tienen aire acondicionado central (CentralAir = Y) y las que no (CentralAir = N).

¿Cómo se interpreta?

Caja (rectángulo): representa el 50 % central de los precios, entre el primer cuartil (Q1) y el tercer cuartil (Q3).

Línea dentro de la caja: representa la mediana del precio.

Bigotes: muestran el rango de valores considerados normales según el criterio del boxplot.

CentralAir = N (sin aire acondicionado central):
