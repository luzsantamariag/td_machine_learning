# Los conceptos básicos: figuras, subgráficos (ejes) y ejes

Matplotlib es un proyecto extenso y puede parecer abrumador al principio. Sin embargo, al familiarizarse con sus componentes, se vuelve mucho más sencillo y accesible.

**Anatomía de un "gráfico"**

El término "gráfico" se usa para referirse a muchas cosas diferentes. Aquí, utilizaremos una terminología consistente (que se refleja en los nombres de las clases subyacentes, etc.):

<img src="./figure_axes_axis_labeled.png" height='450px' width='450px'>


La "Figura" es el contenedor de nivel superior en esta jerarquía. Es la ventana/página principal donde se dibuja todo. Se pueden tener varias figuras independientes y una "Figura" puede contener varios "Ejes".

La mayoría de los gráficos se crean con relación a un eje (Axes) o un subgráfico (Subplot). Un subgráfico se compone de objetos de eje (Axis) y otros elementos. Un subgráfico debe pertenecer a una figura (Figure) (y solo a una). La mayoría de los comandos que utilizaremos se refieren a este subgráfico.

Cada eje tiene un eje X y un eje Y. Estos contienen las marcas, su ubicación, las etiquetas, etc. Controlaremos principalmente las marcas, sus etiquetas y los límites de datos mediante otros mecanismos, por lo que no profundizaremos demasiado en la parte del eje. Sin embargo, vale la pena mencionar aquí de dónde proviene el término "ejes".

## pyplot

`matplotlib.pyplot` es una colección de funciones de estilo comando que hacen que matplotlib funcione como MATLAB. Cada función pyplot modifica una figura: por ejemplo, crea una figura, crea un área de trazado en una figura, traza líneas en un área de trazado, decora el gráfico con etiquetas, etc. En matplotlib.pyplot, se conservan varios estados entre las llamadas a funciones, de modo que se mantiene un registro de elementos como la figura y el área de trazado actuales, y las funciones de trazado se dirigen al subgráfico actual.

```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1, 2, 3, 2.5])
plt.ylabel('algunos números')
```

`plot()` es un comando versátil y acepta un número arbitrario de argumentos. Por ejemplo, para graficar x en función de y, puede usar el comando:

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16]);
```

Utilizando la interfaz pyplot, se construye un gráfico llamando a una secuencia de funciones, y todas ellas se aplican al *subgráfico actual*, de la siguiente manera:

```python
plt.plot([1, 2, 3, 4], [10, 20, 25, 30], color='lightblue', linewidth=3)
plt.scatter([0.3, 3.8, 1.2, 2.5], [11, 25, 9, 26], color='darkgreen', marker='^')
plt.xlim(0.5, 4.5)
plt.title("Título del gráfico")
plt.xlabel("Esta es la etiqueta-x")
plt.ylabel("Esta es la etiqueta-y")
```

Cuando se trabaja con un solo subgráfico en la figura, generalmente está bien trabajar con la interfaz pyplot; sin embargo, cuando se realizan gráficos más complejos o se trabaja dentro de scripts más grandes, se debe pasar explícitamente el objeto *Subplot (Axes)* y/o *Figure* para operar con él.

```python
import numpy as np
def f(t):
    return np.exp(-t) * np.cos(2 * np.pi * t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)
```

```python
t1
```

```python
t2
```

```python
plt.figure()                # Crea una nueva ventana o figura donde se dibujarán las gráficas.
plt.subplot(2, 1, 1)        # Divide la figura en una cuadrícula de: 2 filas, 1 columna, selecciona el primer espacio
plt.plot(t1, f(t1), 'bo')   # Dibuja la primera gráfica: t1 valores del eje X, f(t1) valores del eje Y, b = blue, o = círculos.

plt.subplot(2, 1, 2)        # Vuelve a dividir la figura en 2 filas y 1 columna, pero ahora selecciona el espacio número 2.
plt.plot(t2, np.cos(2 * np.pi * t2), 'r--') # Dibuja la segunda gráfica: t2 eje x, cos... eje y, r = red, -- línea discontinua

plt.ylabel("etiqueta Y")    # ¿Qué subgráfico está modificando esta función?
```

La ventaja de mantener muy claro en nuestro código con qué subgráfico estamos trabajando se hace evidente cuando empezamos a tener varios ejes en una misma figura. Dado que (según el principio zen de Python) **lo explícito es mejor que lo implícito**, recomiendo trabajar con la interfaz orientada a objetos.

## Interfaz orientada a objetos

Al graficar con la interfaz orientada a objetos, se crean los objetos y luego se llaman métodos para cada uno de ellos, realizando cambios específicos. La función `plt.subplots` crea simultáneamente un objeto de figura y un objeto de subgráfico (ejes). Posteriormente, podemos crear un gráfico llamando a los métodos de dichos objetos.

```python
x = np.linspace(-4, 4)      # Por defecto se crean 50 puntos entre -4 y 4
y1, y2 = x**2, x**3         # y1 eleva x al cuadrado, y2 eleva x al cubo
```

```python
x
```

```python
y1
```

```python
y2
```

```python
# Interfaz orientada a objetos
fig, ax = plt.subplots()        # fig representa la figura completa (lienzo), ax  representa el área donde se dibuja el gráfico (los ejes).
ax.plot(x, y1, 'red')           # Dibuja una curva utilizando: x eje X, y1 eje Y, 'red' color rojo.
ax.plot(x, y2, 'blue')          # Dibuja una curva utilizando: x eje X, y2 eje Y, 'blue' color azul.
ax.set_title('Potencias cuadráticas y cúbicas')     # Establece el título del gráfico
ax.set_xlabel('valores x')      # Coloca una etiqueta en el eje X
ax.set_ylabel('valores y')      # Coloca una etiqueta en el eje Y
```

Curiosamente, casi todos los métodos de un objeto *Axes* existen como funciones en el módulo *pyplot* (y viceversa). Por ejemplo, al llamar a `plt.xlabel("x label")`, *pyplot* llama a `ax.set_xlabel("x label")` en el *Axes* que esté "actual". Aquí hay una versión equivalente del ejemplo anterior usando solo pyplot.

```python
# pyplot
plt.plot(x, y1, 'red', x, y2, 'blue')
plt.title('Potencias cuadráticas y cúbicas')
plt.xlabel('valores x')
plt.ylabel('valores y')
```

La utilidad de la Interfaz Orientada a Objetos (IOO) se hace más evidente al trabajar con múltiples subgráficos en una misma figura. Ya hemos mencionado que una figura puede tener más de un eje. Si desea que sus ejes estén en una cuadrícula regular, lo más sencillo es usar `plt.subplots(...)` para crear una figura y agregarle los ejes automáticamente.

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
```

`plt.subplots(...)` creó una nueva figura y le agregó 4 subgráficos. El objeto `axes` que se devolvió es una matriz de objetos NumPy bidimensional. Cada elemento de la matriz es uno de los subgráficos. Están dispuestos como se ven en la figura. Por lo tanto, cuando queremos trabajar con uno de estos ejes, podemos indexar la matriz `axes` y usar los métodos de ese elemento. Por ejemplo:

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
axes[0,0].set_title('Arriba a la izquierda')
axes[0,1].set_title('Arriba a la derecha')
axes[1,0].set_title('Abajo a la izquierda')
axes[1,1].set_title('Abajo a la derecha')

# Para iterar sobre todos los elementos de una matriz numpy multidimensional, utilice el atributo `flat`.
for ax in axes.flat:
    # Eliminar todas las marcas de los ejes x e y...
    ax.set(xticks=[], yticks=[])

fig.tight_layout()  # Revisa cuánto espacio necesita cada elemento y reorganiza los márgenes.
```

### Ejemplo: Crear una figura que muestre la forma de las funciones potenciales.

```python
x = np.linspace(start=-5, stop=5, num=150)
```

```python
# Todas las funciones en un solo eje
fig, ax = plt.subplots(figsize = (7, 4))            # 7 pulgadas a lo ancho, 4 pulgadas a lo alto
ax.plot(x, x, label='Lineal')
ax.plot(x, x**2, label='Cuadrática')
ax.plot(x, x**3, label='Cúbica')
ax.plot(x, x**4, label='A la cuarta potencia')
ax.legend()
```

```python
# 4 ejes/subgráficos: Una función en un eje
fig, axes = plt.subplots(nrows=2, ncols=2, figsize = (7, 4.5))
axes[0,0].set_title('Lineal')
axes[0,0].plot(x, x)
axes[0,1].set_title('Cuadrática')
axes[0,1].plot(x, x**2)
axes[1,0].set_title('Cúbica')
axes[1,0].plot(x, x**3)
axes[1,1].set_title('A la cuarta potencia')
axes[1,1].plot(x, x**4)
fig.tight_layout()
```

¿Qué pasaría si quisiéramos echar un vistazo a las primeras 10 potencias de x?

```python
# Un enfoque más elegante
fig, axes = plt.subplots(nrows=2, ncols=5, figsize = (14, 4.5))
for i, ax in enumerate(axes.flatten()):
    ax.set_title("x a la {:d}".format(i+1))
    ax.plot(x, x**(i+1))
    
fig.tight_layout();
```

# Personalización común: colores, etiquetas, marcas de verificación, límites y anotaciones.

Ahora veremos cómo modificar algunos elementos de un gráfico. Matplotlib permite personalizar casi todos los elementos de un gráfico. Sin embargo, hay algunos elementos que modificamos con mayor frecuencia, y esos son los que analizaremos aquí.

```python
# Generación de datos
x = np.linspace(-np.pi, np.pi, 200)
sine, cosine = np.sin(x), np.cos(x)
```

En esta sección partiremos de este sencillo gráfico:

```python
fig, ax = plt.subplots()
ax.plot(x, sine, x, cosine)
```

a esta gráfica:

```python
fig, ax = plt.subplots()
ax.plot(x, sine, color='red', label='Seno')
ax.plot(x, cosine, color='#165181', label='Coseno')

ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-1.2, 1.2)

ax.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax.set_yticks([-1, 0, 1])

ax.set_xticklabels([r'$-\pi$', r'$-\pi/2$', r'$0$', r'$+\pi/2$', r'$+\pi$'], size=17)
ax.set_yticklabels(['-1','0','+1'], size=17)

ax.legend(loc='upper left')

ax.text(-0.25,0,'(0,0)')                # coord-x, coord-y, 
ax.text(np.pi-0.2,0.05, r'$(\pi,0)$', size=15)

ax.annotate('Origen',
    xy=(0, 0),                          # donde la flecha apunta
    xytext=(1, -0.7),                   # ubicación del texto
    arrowprops=dict(facecolor='blue')
)

ax.axhline(0, color='black', alpha=0.9) # línea horizontal
ax.axvline(0, color='black', alpha=0.9) # línea vertical
ax.grid()
```

## Colores
En las secciones anteriores, aprendiste cómo Matplotlib organiza la creación de gráficos mediante figuras y ejes. Desglosamos los componentes de una figura básica y aprendimos a crearlos. Esta sección repasará algunas de las propiedades que se utilizan en toda la biblioteca, especialmente aquellas que se usan con mayor frecuencia.

Dado que Matplotlib es una biblioteca de gráficos, los colores están asociados a todo lo que se representa en tus figuras. Matplotlib admite un lenguaje muy robusto para especificar colores, que resultará familiar para una amplia variedad de usuarios.

### Nombres de colores
En primer lugar, los colores se pueden especificar como cadenas de texto. Para colores muy básicos, incluso puedes usar una sola letra:

- b: azul
- g: verde
- r: rojo
- c: cian
- m: magenta
- y: amarillo
- k: negro
- w: blanco

Otros nombres de color permitidos son los de HTML/CSS, como "burlywood" y "chartreuse". Consulta la [lista completa](http://www.w3schools.com/html/html_colornames.asp) de los 147 nombres de color.

### Valores hexadecimales
Los colores también se pueden especificar proporcionando una cadena hexadecimal HTML/CSS, como `'#0000FF'` para el azul.

### 256 tonos de gris
Se puede indicar un nivel de gris en lugar de un color pasando una cadena que represente un número entre 0 y 1, ambos inclusive. `'0.0'` es negro, mientras que `'1.0'` es blanco. `'0.75'` sería un tono de gris más claro.

### Tuplas RGB[A]
Es posible que te encuentres con casos en los que las formas anteriores de especificar colores no funcionen. Esto puede ocurrir a veces en niveles más profundos y complejos del código. Cuando todo lo demás falla, el lenguaje universal de colores para matplotlib es la tupla RGB[A]. Esta es la tupla "Rojo", "Verde", "Azul" y, a veces, "Alfa", compuesta por valores flotantes en el rango [0, 1]. El valor 1 indica la saturación completa de ese canal; por lo tanto, una tupla RGBA roja sería `(1.0, 0.0, 0.0, 1.0)`, mientras que una tupla RGBA verde parcialmente transparente sería `(0.0, 1.0, 0.0, 0.75)`. La documentación generalmente especificará si acepta tuplas RGB o RGBA. En ocasiones, se requiere una lista de tuplas para múltiples colores, e incluso se puede proporcionar una matriz NumPy de Nx3 o Nx4 en estos casos.

En funciones como `plot()` y `scatter()`, aunque parezca que aceptan una especificación de color, lo que realmente necesitan es una "especificación de formato", que incluye el color como parte del formato. Desafortunadamente, dichas especificaciones solo admiten cadenas de texto, por lo que las tuplas RGB[A] no son compatibles con estos argumentos (pero sí se puede pasar una tupla RGB[A] como argumento de "color").

Cabe destacar que, a menudo, existe un argumento independiente para "alfa" dondequiera que se pueda especificar un color. El valor de "alfa" generalmente tendrá prioridad sobre el valor alfa en la tupla RGBA. No hay una solución sencilla para este problema.

## Gráficos de líneas

Un gráfico de líneas es un tipo de gráfico que muestra información como una serie de puntos de datos llamados "marcadores", conectados por segmentos de línea recta. Este tipo de gráfico se utiliza para representar datos de series temporales, funciones matemáticas o relaciones continuas entre variables.

**Estilos de línea**: se utilizan comúnmente para distinguir entre líneas. Existen algunos estilos de línea predefinidos disponibles. Cabe destacar que existen técnicas avanzadas para especificar estilos de línea personalizados. [Aquí](http://matplotlib.org/1.3.0/examples/lines_bars_and_markers/line_demo_dash_control.html) se muestra un ejemplo de un patrón de guiones personalizado.

estilo de línea    | descripción
-------------------|------------------------------
'-'                | sólido
'--'               | Discontinua
'-.'               | línea de puntos y rayas
':'                | punteado
'None'             | no dibuja nada
' '                | no dibuja nada
''                 | no dibuja nada

**Otros atributos**: con casi cualquier gráfico que puedas crear, hay muchos atributos que se pueden modificar para que las líneas y los marcadores se ajusten a tus necesidades. Ten en cuenta que para muchas funciones de trazado, matplotlib alternará los colores para cada conjunto de datos que traces. ## Gráficos de barras

Un gráfico de barras es una representación gráfica que muestra datos mediante barras rectangulares cuya longitud es proporcional a los valores que representan. Las barras pueden trazarse vertical u horizontalmente.

Este tipo de gráfico también se puede utilizar para mostrar comparaciones entre categorías. Un eje del gráfico muestra las categorías específicas que se comparan, y el otro eje representa un valor discreto. Algunos gráficos de barras presentan barras agrupadas en conjuntos de más de una.

```python
# Los 10 mejores países en los Juegos Olímpicos de Río
countries = ['USA','GBR','CHN','RUS','GER','JPN','FRA','KOR','ITA','AUS']
gold =   [46, 27, 26, 19, 17, 12, 10, 9,  8,  8]
silver = [37, 23, 18, 18, 10,  8, 18, 3, 12, 11]
bronze = [38, 17, 26, 19, 15, 21, 14, 9,  8, 10]
```

```python
fig, ax = plt.subplots(figsize=(10,5.5))
ax.bar(np.arange(10), gold, color="#FFDF00", width=0.4, label='Oro')
ax.bar(np.arange(10) + 0.4, silver, color="#C0C0C0", width=0.4, label='Plata')

ax.set_xticks(np.arange(0.2, 10, 1))
ax.set_xticklabels(countries)

for x,g,s in zip(np.arange(10), gold, silver):
    ax.text(x - 0.1, g + 0.5, g)        # anotando los oros
    ax.text(x + 0.3, s + 0.5, s)        # anotando las platas

ax.set_title('Medallas de oro y plata en Río', size=16)
ax.set_xlabel('Pais', size=14)
ax.set_ylabel('Número de medallas', size=14)
ax.legend(loc='upper right')
```

## Histogramas

Un histograma es una representación gráfica de la distribución de datos numéricos. Es una estimación de la distribución de probabilidad de una variable continua (variable cuantitativa). Se trata de un tipo de gráfico de barras. Para construir un histograma, el primer paso es agrupar los valores en intervalos, es decir, dividir todo el rango de valores en una serie de intervalos y, a continuación, contar cuántos valores caen en cada intervalo. Los intervalos suelen especificarse como intervalos consecutivos y no superpuestos de una variable. Los intervalos deben ser adyacentes y, a menudo (aunque no es obligatorio), tienen el mismo tamaño.

Los histogramas ofrecen una idea aproximada de la densidad de la distribución subyacente de los datos.

```python
iqs = np.random.normal(loc=100, scale=10, size=300)
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(12,4))

ax[0].hist(iqs, bins=18, edgecolor='k')
ax[0].set_title('Histograma de frecuencia')
    
ax[1].hist(iqs, bins=18, color = 'red', edgecolor='k', density=True)
ax[1].set_title('Histograma de densidad (normalizado)')
```

## Diagrama de dispersión

Un diagrama de dispersión es un tipo de gráfico o diagrama matemático que utiliza coordenadas cartesianas para mostrar valores de dos variables en un conjunto de datos. Si los puntos están codificados por colores, se puede mostrar una variable adicional. Los datos se representan como una colección de puntos, donde el valor de una variable determina su posición en el eje horizontal y el valor de la otra variable determina su posición en el eje vertical.

```python
fig, ax = plt.subplots()
ax.scatter(gold, silver, marker='o')

ax.set_title('Oro contra plata en los Juegos Olímpicos de Río', size=16)
ax.set_xlabel('Medallas de oro', size=14)
ax.set_ylabel('Medallas de plata', size=14);
```
