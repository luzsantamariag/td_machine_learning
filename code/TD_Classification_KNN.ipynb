En cualquier iniciativa de transformación digital, la calidad y disponibilidad de los datos gobiernan el proceso analítico. `scikit-learn` provee repositorios de prueba estandarizados que facilitan la experimentación y validación de modelos para problemas representativos de negocio, como clasificación, regresión y agrupamiento.

```python
# Importación de bibliotecas base del stack analítico
import numpy as np
import pandas as pd
from sklearn import datasets
```

### Carga del Conjunto de Datos
- El dataset IRIS es probablemente el primer conjunto de datos que los estudiantes y principiantes experimentan mientras aprenden ciencia de datos y *machine learning*. El dataset IRIS es simple y fácil de entender, ha sido utilizado por estudiantes, investigadores y profesionales durante décadas para aprender los conceptos básicos de clasificación y análisis de datos. 

- Fue publicado en 1936 por el estadístico británico Ronald A. Fisher en el artículo “**El uso de mediciones múltiples en problemas taxonómicos**” y desde entonces, el dataset IRIS se ha hecho conocido como el “Hola mundo” del ML porque proporciona una introducción muy efectiva a los datasets y algoritmos del mundo real para los interesados.

<img src="./iris.png" width=600>

> El conjunto de datos Iris tiene **150 filas** y **4 características** medidas (en centímetros) para cada flor:
Estas son las medidas que utilizamos para hacer nuestra predicción. Puedes pensar en ellos como las piezas de un rompecabezas.
- `sepal length (cm)` Longitud del sépalo (cm): El sépalo es la parte de la flor que protege el brote antes de florecer. Esta característica es una medida de la longitud del sépalo. 
- `sepal width (cm)` Ancho del sépalo (cm): es una medida del ancho del sépalo.
- `petal length (cm)` Longitud del pétalo (cm): El pétalo es la parte muy colorida de la flor. Esta es una medida de cuánto tiempo dura. 
- `petal width (cm)` Ancho del pétalo (cm): es una medida del ancho del pétalo. 

> Las clases objetivo (los resultados): 
- `Iris Setosa (0)`: Esta especie se distingue claramente de las otras dos especies de Iris, por lo que puede considerarse como el "modo fácil" de este conjunto de datos para un clasificador. 
- `Iris Versicolor (1)`: Esta especie es muy similar a Virginica, lo que significa que es más difícil distinguirlas. ¡Ahí radica la verdadera dificultad!
- `Iris Virginica (2)`: La tercera especie que es más probable que se confunda con Versicolor basándose únicamente en las medidas de dimensiones. 

> Se carga un conjunto estructurado que servirá de marco conceptual para abordar problemas de **clasificación multiclase** (asignación automatizada a una de tres categorías estratégicas a partir de variables cuantitativas).

```python
# Carga del dataset en un DataFrame de Pandas
iris = datasets.load_iris(as_frame=True)

# X: Matriz de características / variables independientes (Features)
X = iris.data

# y: Vector de etiquetas / variable objetivo de negocio (Target)
y = iris.target

# Visualización del esquema de datos para gobernanza y análisis exploratorio de datos (EDA)
iris_df = iris.frame
iris_df['categoria_nombre'] = iris.target_names[iris.target]
iris_df.head()
```

```python
# Clases o categorías objetivo a predecir por la organización
for idx, name in enumerate(iris.target_names):
    print(f"Clase {idx}: {name}")
```

## 2. Paradigma del Modelo: Problema de Clasificación Supervisada

> Problema de **Clasificación**: Cuando el objetivo estratégico consiste en catalogar, diagnosticar o categorizar una entidad o clase previamente definida (por ejemplo, para identificar clientes con riesgo de deserción (churn), detectar transacciones potencialmente fraudulentas o clasificar perfiles según su nivel de adopción tecnológica) a partir de los datos históricos. 

> En un problema de aprendizaje supervisado, el modelo aprende la relación entre las características o variables de entrada y una variable objetivo conocida a partir de ejemplos históricos. Posteriormente, utiliza ese aprendizaje para clasificar nuevos casos cuya categoría aún no se conoce.

> **Referencia estratégica:** El [mapa de algoritmos de Aprendizaje Supervisado de Scikit-Learn](https://scikit-learn.org/stable/supervised_learning.html) orienta a los involucrados en la selección del algoritmo e clasificación adecuado según el volumen de datos y el tipo de resultado esperado. 

## 3. Arquitectura de Scikit-Learn: La API del Estimador (*Estimator Pattern*)

Scikit-Learn estandariza el flujo de trabajo de la machine learning mediante el patrón **Estimador**. Todo algoritmo expone una interfaz homogénea orientada a facilitar la interoperabilidad en la arquitectura de software:

1. **Instanciación:** Configuración de hiperparámetros del modelo.
2. **Ajuste (`fit`):** Proceso de entrenamiento o aprendizaje a partir de la evidencia empírica.
3. **Inferencia (`predict`):** Despliegue del conocimiento aprendido sobre nuevos casos del entorno de negocio.

> **Referencia estratégica:** El [mapa de algoritmos de Scikit-Learn](https://scikit-learn.org/stable/tutorial/machine_learning_map/) orienta a los involucrados en la selección del modelo adecuado según el volumen de datos y el tipo de resultado esperado.

### 3.1: Selección e Importación del Estimador
Seleccionamos el algoritmo **K-Nearest Neighbors (KNN)**, un modelo basado en similitud que clasifica nuevos eventos analizando el comportamiento de sus vecinos más cercanos.

```python
from sklearn.neighbors import KNeighborsClassifier
```

### Configuración del Modelo (Instanciación)
Definimos los parámetros que guiarán la toma de decisiones del algoritmo (por ejemplo, considerar los $k=3$ patrones históricos más similares).

```python
clasificador_knn = KNeighborsClassifier(n_neighbors=3)
```

## 4. Partición Estratégica: Entrenamiento vs. Validación (*Train/Test Split*)

En la gestión de la transformación digital, un error crítico es evaluar la efectividad de una solución analítica sobre los mismos datos utilizados para calibrarla (lo que induce a *sobreajuste* o falso optimismo).

Separamos el conjunto de datos en:
* **Entrenamiento (70%):** Información para que el algoritmo aprenda los patrones.
* **Prueba (30%):** Datos no observados para validar la capacidad real de generalización en producción.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

print(f"Casos para entrenamiento: {X_train.shape[0]}")
print(f"Casos para validación de negocio: {X_test.shape[0]}")
```

### Entrenamiento del Modelo (`fit`)
Aquí confrontamos al algoritmo con los datos de entrenamiento para que construya su espacio de vecindad.

```python
# Ejecución del entrenamiento sobre el set de calibración
clasificador_knn.fit(X_train, y_train)
```

## 5. Evaluación de Desempeño y Métricas de Negocio

Antes de aprobar la integración de un modelo en la cadena de valor o en los procesos operativos de la organización, la dirección debe contrastar el nivel de acierto (*Accuracy*) y la consistencia de las predicciones.

> **Referencia estratégica:** El [Model selection and evaluations - Scikit-Learn](https://scikit-learn.org/stable/model_selection.html) orienta a los involucrados en la selección y evaluación de los modelos o algoritmos de Machine Learning.

```python
from sklearn.metrics import accuracy_score, classification_report

# Generar predicciones sobre el conjunto de prueba no observado
y_pred = clasificador_knn.predict(X_test)

# Evaluación del nivel de precisión global
exactitud = accuracy_score(y_test, y_pred)
print(f"Nivel de acierto global (Accuracy): {exactitud:.2%}\n")
print("Reporte analítico por categoría:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```

### 5.1. Evaluación Robusta del Rendimiento: Validación Cruzada (*Cross-Validation*)

> En lugar de confiar en una única partición que puede estar sesgada, la **Validación Cruzada en K Pliegues (K-Fold Cross-Validation)** divide los datos en $K$ subconjuntos e itera entrenando en $K-1$ y evaluando en el restante. Esto proporciona una estimación insesgada de la estabilidad del modelo frente a la variabilidad del entorno operacional.

> **Referencia estratégica:** El [Cross-validation: evaluating estimator performance - Scikit-Learn](https://scikit-learn.org/stable/modules/cross_validation.html) Aprender a evaluar el desempeño con validación cruzada en el entrenamiento de modelos.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Estimador base de referencia
modelo_base = KNeighborsClassifier(n_neighbors=5)

# Configuración de validación cruzada estratificada de 5 pliegues
cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Evaluación con Cross-Validation
scores_cv = cross_val_score(modelo_base, X_train, y_train, cv=cv_estratificado, scoring='accuracy')

print(f"Rendimientos por cada pliegue: {np.round(scores_cv, 4)}")
print(f"Exactitud media esperada (Cross-Validation Mean): {scores_cv.mean():.2%}")
print(f"Desviación estándar (estabilidad del modelo): ±{scores_cv.std():.4f}")
```

# 6. Selección y Optimización de Hiperparámetros
> En Machine Learning, los **hiperparámetros** son configuraciones que controlan el comportamiento de un algoritmo antes de iniciar su entrenamiento.
> Su optimización busca identificar una configuración que permita obtener un buen desempeño predictivo, manteniendo un equilibrio entre:

- **Desempeño del modelo**
- **Costo computacional**
- **Tiempo de experimentación**
- **Complejidad operativa**
- **Reproducibilidad**

> En `scikit-learn` existen dos estrategias ampliamente utilizadas para realizar esta búsqueda:
1. **Grid Search CV:** evalúa exhaustivamente todas las combinaciones definidas.
2. **Randomized Search CV:** evalúa una muestra aleatoria de configuraciones dentro del espacio de búsqueda.

> **Referencia estratégica:** El [Tuning the hyper-parameters of an estimator - Scikit-Learn](https://scikit-learn.org/stable/modules/grid_search.html) Aprender a ajustar los hiperparámetros de los modelos de Machine Learning.

### 6.1 Optimización Exhaustiva de Hiperparámetros: Grid Search CV
> **GridSearchCV** efectúa un barrido exhaustivo por todas las combinaciones posibles de hiperparámetros dentro de un espacio de búsqueda acotado. Cada combinación se evalúa con validación cruzada interna.
> * **Ventaja:** Garantiza encontrar la mejor combinación dentro de la cuadrícula evaluada.
> * **Trade-off estratégico:** Mayor costo computacional y tiempo de cómputo en espacios de búsqueda extensos.

```python
from sklearn.model_selection import GridSearchCV

# Definición del espacio de búsqueda para KNeighborsClassifier
param_grid = {
    'n_neighbors': list(range(1, 21)),
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski']
}

# Configuración de GridSearchCV con 5 pliegues cruzados
grid_search = GridSearchCV(
    estimator=KNeighborsClassifier(),
    param_grid=param_grid,
    cv=cv_estratificado,
    scoring='accuracy',
    n_jobs=-1
)

# Ejecución del ajuste y búsqueda
grid_search.fit(X_train, y_train)

print("Resultados del método de selección Search CV")
print(f"Mejores hiperparámetros encontrados: {grid_search.best_params_}")
print(f"Mejor exactitud media validada en CV: {grid_search.best_score_:.2%}")
```

### 6.2. Optimización Eficiente en Espacios Complejos: Randomized Search CV
> **RandomizedSearchCV** muestrea un número fijo de combinaciones aleatorias (`n_iter`) dentro de distribuciones o rangos continuos y discretos.
> * **Ventaja:** Permite explorar espacios hiperparamétricos considerablemente mayores con un presupuesto computacional controlado.
> * **Aplicabilidad en Transformación Digital:** Ideal para proyectos ágiles donde se requiere balancear el tiempo de experimentación (*time-to-market*) y la precisión del modelo.

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

# Espacio de búsqueda probabilístico
param_dist = {
    'n_neighbors': randint(1, 30),
    'weights': ['uniform', 'distance'],
    'p': [1, 2]  # 1: manhattan, 2: euclidean
}

# Configuración de RandomizedSearchCV muestreando 15 combinaciones
random_search = RandomizedSearchCV(
    estimator=KNeighborsClassifier(),
    param_distributions=param_dist,
    n_iter=15,
    cv=cv_estratificado,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1
)

# Ejecución de la búsqueda aleatorizada
random_search.fit(X_train, y_train)

print("Resultados del método de selección Randomized Search CV")
print(f"Mejores hiperparámetros seleccionados: {random_search.best_params_}")
print(f"Mejor exactitud media validada en CV: {random_search.best_score_:.2%}")
```

### 6.3 Selección y Evaluación Final del Mejor Modelo en Producción

> Seleccionamos el estimador óptimo derivado de la búsqueda y evaluamos su desempeño definitivo sobre el conjunto de prueba independiente (`X_test`), simulando las condiciones reales de despliegue.

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Despliegue del mejor modelo obtenido (GridSearchCV refit=True por defecto)
mejor_modelo = grid_search.best_estimator_

# Predicciones sobre datos no observados
y_pred_final = mejor_modelo.predict(X_test)

exactitud_final = accuracy_score(y_test, y_pred_final)
print(f"Exactitud final en datos de prueba no observados: {exactitud_final:.2%}\n")
print("Reporte Ejecutivo de Clasificación:")
print(classification_report(y_test, y_pred_final, target_names=iris.target_names))

print("Matriz de Confusión:")
print(confusion_matrix(y_test, y_pred_final))
```

## 7. Despliegue e Inferencia en Escenarios Operativos Reales

Esto permite mostrar ante la gerencia cómo un modelo entrenado puede recibir nuevos eventos de producción y generar una categorización automática.

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# Simulación de 10 nuevos eventos / registros capturados en producción (Matriz 2D)
nuevos_casos = np.array([
    [5.0, 3.4, 1.5, 0.2],  # Perfil operacional caso A
    [5.4, 3.0, 1.6, 0.2],  # Perfil operacional caso B
    [4.9, 3.1, 1.5, 0.1],  # Perfil operacional caso C
    [5.8, 2.7, 4.1, 1.0],  # Perfil operacional caso D
    [6.1, 2.8, 4.7, 1.2],  # Perfil operacional caso E
    [5.6, 2.9, 3.6, 1.3],  # Perfil operacional caso F
    [6.3, 3.3, 6.0, 2.5],  # Perfil operacional caso G
    [6.5, 3.0, 5.8, 2.2],  # Perfil operacional caso H
    [7.1, 3.0, 5.9, 2.1],  # Perfil operacional caso I
    [6.4, 2.8, 5.6, 2.2]   # Perfil operacional caso J
])

# Inferencia predictiva
predicciones_numericas = clasificador_knn.predict(nuevos_casos)
predicciones_etiquetas = [iris.target_names[i] for i in predicciones_numericas]

for i, (caso, etiqueta) in enumerate(zip(nuevos_casos, predicciones_etiquetas), start=1):
    print(f"Caso Operativo {i} {caso} --> Categorización sugerida: {etiqueta}")
```
