# Entornos de Transformación Digital: Análisis Exploratorio (EDA), Análisis Confirmatorio (CDA) e Ingeniería de Características
---

> En los ecosistemas corporativos modernos, la madurez analítica de una organización depende de su capacidad para transformar datos operacionales sin procesar o crudos (`data raw`) en decisiones estratégicas fundamentadas y auditables. En los procesos de transformación digital, este trayecto se consolida a través de tres fases complementarias e indispensables:

1. **Análisis Exploratorio de Datos (*Exploratory Data Analysis - EDA*):** Espacio de descubrimiento y formulación de hipótesis. Permite a los líderes comprender las distribuciones, variabilidad, valores extremos y relaciones latentes en los datos del negocio sin suposiciones a priori.
2. **Análisis Confirmatorio de Datos (*Confirmatory Data Analysis - CDA*):** Marco formal de contrastación y gobernanza estadística que evalúa si los patrones observados en el EDA son estadísticamente significativos o producto del azar, empleando pruebas de hipótesis ($t$-Student, ANOVA, Chi-cuadrado e intervalos de confianza) requeridas por comités directivos y regulatorios.
3. **Ingeniería de Características Estratégica (*Feature Engineering*):** Creación y optimización de variables de negocio (antigüedad, superficie total, ratios operacionales), transformaciones para estabilizar la varianza (logaritmo sobre precios sesgados) y selección metódica de predictores (Filtro, Embebidos, Envoltura y PCA) para reducir costos computacionales y maximizar el retorno de inversión (*ROI*).

## 2. Análisis Exploratorio de Datos (EDA) - Perspectiva de Negocio

### 2.1 Diagnóstico de la Métrica Principal de Desempeño: `SalePrice`
> En finanzas y valoración de activos la verificación si la variable objetivo presenta una distribución normal o asimétrica determina las decisiones de modelado y precio `SalePrice`. Para tal propósito, se utilizan las siguientes `medidas estadísticas` que describen la forma de la distribución de estos datos:
1. `Skewness` (asimetría): indica si los datos están distribuidos de manera simétrica o si tienen una cola más larga hacia un lado (Izquierdo -1, centro: 0 y derecho: 1). Muchas personas pueden tener ingresos relativamente bajos/medios, mientras que unas pocas, tienen ingresos extremadamente altos. Esos valores altos generan una cola hacia la derecha.
2. `Kurtosis` (curtosis): mide principalmente qué tan pesadas son las colas de una distribución, es decir, qué tan propensos son los datos a presentar valores extremos `outliers`.

````python
precio = housing['SalePrice']
media = precio.mean()
mediana = precio.median()
skewness = precio.skew()
kurtosis = precio.kurt()

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(precio, kde=True, color='#1f77b4', ax=ax[0])
ax[0].axvline(media, color='red', linestyle='--', label=f'Media: ${media:,.0f}')
ax[0].axvline(mediana, color='green', linestyle='-', label=f'Mediana: ${mediana:,.0f}')
ax[0].set_title('Distribución Asimétrica de Precios de Venta (EDA)')
ax[0].legend()

stats.probplot(precio, dist="norm", plot=ax[1])
ax[1].set_title('Gráfico Q-Q Normal (Evidencia de Colas Pesadas)')

plt.tight_layout()
plt.show()

print(f"Diagnóstico Estadístico del Precio de Venta:")
print(f"- Asimetría (Skewness): {skewness:.3f} (Asimetría positiva notable a la derecha)")
print(f"- Curtosis (Excess Kurtosis): {kurtosis:.3f} (Colas pesadas y presencia de outliers de alta gama)")
````

## 3. Análisis Confirmatorio de Datos (CDA) - Rigor Estadístico para los Directivos

- El **CDA** permite evaluar formalmente afirmaciones sobre una población utilizando los datos de una muestra. Para ello se utiliza un **contraste de hipótesis**, que sigue una serie de pasos.

> **1. Planteamiento de las hipótesis**: Comienza estableciendo dos hipótesis sobre un parámetro de la población.
- **Hipótesis nula (`H₀`)**: representa la afirmación que se considera cierta inicialmente, con un valor específico del parámetro.
- **Hipótesis alternativa (`Hₐ`)**: representa la afirmación que desea investigar y contempla valores diferentes a los establecidos por `H₀`.

> **2. Establecimiento del nivel de significancia `α` (alpha)**.
- Este valor representa el umbral utilizado para decidir sobre la `Hₐ` y se relaciona con la probabilidad de cometer un **error de tipo I**, es decir, rechazar `H₀` cuando realmente es cierta.
- Los valores más utilizados son: `α = 0.05` → 5%, `α = 0.01` → 1% y `α = 0.10` → 10%. 

> **3. Cálculo del estadístico de prueba y del p-value** a partir de los datos de la muestra se calcula un **estadístico de prueba**.

 | Prueba | ¿Cuándo usarla? | ¿Qué pregunta responde? | Ejemplo en House Prices | Hipótesis nula (H₀) | Interpretación para la gestión |
|---|---|---|---|---|---|
| **t de Student – una muestra** | Comparar el promedio de una variable con un valor de referencia | ¿El promedio observado es diferente del valor esperado? | ¿El **SalePrice** promedio de las viviendas es diferente de **$200.000**? | El promedio de SalePrice = $200.000 | Evaluar si el precio promedio del mercado se encuentra por encima o por debajo de un valor de referencia. |
| **t de Student – dos muestras independientes** | Comparar el promedio de dos grupos independientes | ¿Existen diferencias entre los promedios de dos grupos? | ¿El **SalePrice** promedio difiere entre viviendas **con y sin garaje**? | Los promedios de SalePrice son iguales | Identificar diferencias de precio asociadas con una característica de la vivienda. |
| **t pareada (Paired t-test)** | Comparar dos mediciones relacionadas de los mismos elementos | ¿Existe un cambio promedio entre dos mediciones relacionadas? | Comparar dos **valoraciones relacionadas de una misma vivienda**, si el análisis dispone de una medición antes/después. | El cambio promedio = 0 | Evaluar cambios en una misma unidad de análisis cuando existen mediciones emparejadas. |
| **Welch t-test** | Comparar dos grupos cuando sus varianzas pueden ser diferentes o los tamaños de muestra son desiguales | ¿Los promedios de dos grupos son diferentes sin asumir varianzas iguales? | ¿El **SalePrice** promedio difiere entre viviendas con **OverallQual ≤ 5** y viviendas con **OverallQual > 5**? | Los promedios de SalePrice son iguales | Comparar segmentos de viviendas cuando presentan distinta dispersión o tamaño de muestra. |
| **ANOVA de un factor** | Comparar tres o más grupos | ¿Existe alguna diferencia entre los promedios de varios grupos? | ¿El **SalePrice** promedio difiere según las categorías de **OverallQual**? | Todos los promedios de SalePrice son iguales | Identificar si la calidad de la vivienda está asociada con diferencias de precio entre categorías. |
| **ANOVA factorial** | Analizar simultáneamente dos o más factores y su interacción | ¿Dos o más características influyen conjuntamente en la variable analizada? | ¿El **SalePrice** varía según **OverallQual** y la **presencia de garaje**, y existe interacción entre ambos factores? | No existen efectos ni interacción | Analizar cómo diferentes características pueden estar asociadas conjuntamente con el precio de la vivienda. |


### Regla de decisión

> **4. Toma de decisión** Se compara el `p-value` con el nivel de significancia `α`. La regla habitual es:
```text
- Si p-value < α  → Rechazar H₀
- Si p-value ≥ α  → No rechazar H₀

### ¿Por qué el CDA es mandatorio en la Transformación Digital?
Mientras el EDA formula conjeturas visuales ("las casas con aire acondicionado parecen más caras"), el **Análisis Confirmatorio de Datos (CDA)** somete estas intuiciones a pruebas de significancia estadística con hipótesis formales, niveles de confianza y p-valores.

Esto evita que las organizaciones inviertan capital en automatizaciones basadas en correlaciones espurias `Spurious correlations` (La relación entre dos variables puede ser engañosa) o sesgos muestrales `Sampling bias` (la muestra no representa correctamente a la población).

--- 

### Prueba Confirmatoria 1: ¿El Aire Acondicionado genera un incremento de precio estadísticamente significativo?
* **Hipótesis Nula ($H_0$):** La media de precios de venta de inmuebles con aire acondicionado es igual a la de inmuebles sin aire acondicionado ($\mu_{AC} = \mu_{SinAC}$).
* **Hipótesis Alternativa ($H_1$):** La media de precios con aire acondicionado es significativamente mayor ($\mu_{AC} > \mu_{SinAC}$).

````python
precios_con_ac = housing.loc[housing['CentralAir'] == 'Y', 'SalePrice']
precios_sin_ac = housing.loc[housing['CentralAir'] == 'N', 'SalePrice']

# Prueba de Welch t-test (asumiendo varianzas desiguales)
t_stat, p_valor_t = stats.ttest_ind(precios_con_ac, precios_sin_ac, equal_var=False, alternative='greater')

# Intervalo de confianza al 95% para la diferencia de medias
diff_medias = precios_con_ac.mean() - precios_sin_ac.mean()
std_err_diff = np.sqrt(precios_con_ac.var()/len(precios_con_ac) + precios_sin_ac.var()/len(precios_sin_ac))
ci_inf = diff_medias - 1.96 * std_err_diff
ci_sup = diff_medias + 1.96 * std_err_diff

print("Reporte de Análisis Confirmatorio de Datos (CDA): Welch t-test")
print(f"Precio medio con climatización:   ${precios_con_ac.mean():,.2f}")
print(f"Precio medio sin climatización:   ${precios_sin_ac.mean():,.2f}")
print(f"Diferencia promedio de mercado:   ${diff_medias:,.2f}")
print(f"Intervalo de Confianza del 95%:   [${ci_inf:,.2f} , ${ci_sup:,.2f}]")
print(f"Estadístico t: {t_stat:.4f} | p-valor: {p_valor_t:.4e}")

if p_valor_t < 0.05:
    print(">> Conclusión Directiva: Se RECHAZA H0 (p < 0.05). Se confirma formalmente que la dotación de AC aporta una prima de valor cuantificable y no atribuible al azar.")
````

### Prueba Confirmatoria 2: ¿Difieren los Precios Significativamente según la Calidad de los Materiales? (ANOVA de Una Vía)
* **Hipótesis Nula ($H_0$):** Todas las categorías de calidad de materiales (`OverallQual`) tienen el mismo precio medio de mercado.
* **Hipótesis Alternativa ($H_1$):** Al menos un nivel de calidad presenta un precio medio significativamente distinto.

````python
# Agrupamiento de precios según niveles de calidad (OverallQual: 1 a 10)
grupos_calidad = [grupo['SalePrice'].values for _, grupo in housing.groupby('OverallQual')]

f_stat, p_valor_anova = stats.f_oneway(*grupos_calidad)

print("Reporte de Análisis Confirmatorio de Datos (CDA): Anova de una vía")
print(f"Estadístico F de Fisher: {f_stat:.4f}")
print(f"p-valor: {p_valor_anova:.4e}")
if p_valor_anova < 0.05:
    print(">> Conclusión Directiva: Se RECHAZA H0. La percepción de acabados y materiales genera una diferenciación estructural del precio con certeza estadística absoluta.")
````

## 4. Ingeniería de Características Estratégica (*Feature Engineering*)

> La ingeniería de características traduce el conocimiento empírico validado en representaciones matemáticas que maximizan el poder explicativo de los modelos analíticos.

### 4.1 Creación de Características de Negocio Sintéticos (*Feature Extraction*)
> Diseñamos tres métricas estratégicas:
1. **`Age` (Antigüedad al momento de la venta de la propiedad):** $YrSold - YearBuilt$.
2. **`TotalLivingSF` (Espacio Habitable Efectivo Total):** $1stFlrSF + 2ndFlrSF$.
3. **`Has2ndFloor` (Bandera Operacional Binaria):** Identifica si el inmueble posee verticalidad o segundo piso.

````python
# 1. Antigüedad del inmueble en el momento de la transacción
housing['Age'] = housing['YrSold'].astype(float) - housing['YearBuilt'].astype(float)

# 2. Superficie habitable total acumulada
housing['TotalLivingSF'] = housing['1stFlrSF'].astype(float) + housing['2ndFlrSF'].astype(float)

# 3. Indicador de presencia de segunda planta
housing['Has2ndFloor'] = (housing['2ndFlrSF'] > 0).astype(int)

# Filtrado de vecindarios con representatividad estadística (mínimo 30 observaciones)
counts_nb = housing['Neighborhood'].value_counts()
valid_nb = list(counts_nb[counts_nb >= 30].index)
housing = housing[housing['Neighborhood'].isin(valid_nb)].copy()

print("Nuevas variables estratégicas incorporadas al repositorio corporativo:")
print(f"Inventario de Activos Inmobiliarios: {housing.shape[0]} observaciones y {housing.shape[1]} variables recolectadas.")
housing[['Age', 'TotalLivingSF', 'Has2ndFloor', 'SalePrice']].head()
````

### 4.2 Transformación Funcional de la Variable Objetivo (*Variance Stabilization*)
> Como se descubrió en el EDA y se confirmó en el CDA, `SalePrice` está sesgado hacia la derecha. Aplicamos la transformación logarítmica natural $\log(1 + \text{SalePrice})$, la cual estabiliza la varianza residual, aproxima la normalidad y mejora la precisión de los modelos de precio `SalePrice` basados en gradientes y mínimos cuadrados.

````python
housing['Log_SalePrice'] = np.log1p(housing['SalePrice'])

fig, ax = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(housing['Log_SalePrice'], kde=True, color='#27ae60', ax=ax[0])
ax[0].set_title('Precio de Venta Normalizado (Escala Logarítmica)')
stats.probplot(housing['Log_SalePrice'], dist="norm", plot=ax[1])
ax[1].set_title('Gráfico Q-Q Normal tras Transformación Logarítmica')
plt.tight_layout()
plt.show()

print(f"Nueva Asimetría tras transformación: {housing['Log_SalePrice'].skew():.3f} (Distribución simétrica casi ideal)")
print(f"Nueva Curtosis tras transformación:  {housing['Log_SalePrice'].kurt():.3f}")
````

### 4.3 Codificación Categórica y Partición Estratégica: Entrenamiento vs. Validación (*Train/Test Split*)
Codificamos variables cualitativas vía *One-Hot Encoding* y dividimos en conjuntos de entrenamiento (80%) y prueba (20%), evitando cualquier fuga de datos (*Data Leakage*).

````python
from sklearn.model_selection import train_test_split

# Selección de características predictoras refinadas
final_num_vars = ['LotArea', 'OverallQual', 'OverallCond', 'Age', 'TotalLivingSF', 'BedroomAbvGr', 'Has2ndFloor']
final_cat_vars = ['MSZoning', 'LotShape', 'Neighborhood', 'CentralAir', 'SaleCondition']

df_preparado = pd.get_dummies(housing[final_num_vars + final_cat_vars], drop_first=True)
X = df_preparado
y_log = housing['Log_SalePrice']
y_original = housing['SalePrice']
nombres_features = list(X.columns)

# Partición estratégica reproducible
X_train, X_test, y_train_log, y_test_log, y_train_orig, y_test_orig = train_test_split(
    X, y_log, y_original, test_size=0.20, random_state=42
)

print(f"Casos para entrenamiento y selección: {X_train.shape[0]} | Validación: {X_test.shape[0]}")
print(f"Matriz final de variables operacionales: {X_train.shape[1]} columnas.")
````

### 4.4 Selección Sistemática de Características (*Feature Selection Framework*)

Para optimizar la infraestructura de cómputo y el tiempo de respuesta en producción, aplicamos los tres métodos de selección:

1. **Método de Filtro:** `SelectKBest` con F-Regression (prueba ANOVA).
2. **Método Embebido:** Importancia de impureza (MDI) con `RandomForestRegressor`.
3. **Método de Envoltura (*Wrapper*):** Eliminación Recursiva de Características (`RFE`).

> **Referencia estratégica:** El [mapa de algoritmos de Selección de Características de Scikit-Learn](https://scikit-learn.org/stable/api/sklearn.feature_selection.html) orienta a los involucrados en la selección del modelo adecuado según el volumen de datos y el tipo de resultado esperado.

````python
from IPython import display
from sklearn.feature_selection import SelectKBest, f_regression, RFE
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor

# 1. Filtro: SelectKBest (k=10)
selector_f = SelectKBest(score_func=f_regression, k=10)
selector_f.fit(X_train, y_train_log)

# 2. Embebido: Random Forest Importances
rf_selector = RandomForestRegressor(n_estimators=100, random_state=42)
rf_selector.fit(X_train, y_train_log)

# 3. Envoltura: RFE con Árbol Base (k=10)
rfe_selector = RFE(estimator=DecisionTreeRegressor(random_state=42), n_features_to_select=10, step=1)
rfe_selector.fit(X_train, y_train_log)

# Cuadro directivo comparativo de selección
df_seleccion = pd.DataFrame({
    'Variable': nombres_features,
    'F_Score': selector_f.scores_,
    'Importancia_RF': rf_selector.feature_importances_,
    'Ranking_RFE': rfe_selector.ranking_,
    'Seleccionada_RFE': rfe_selector.support_
}).sort_values(by='Importancia_RF', ascending=False)

print("Top 10 variables operacionales de mayor impacto en el modelo:")
display.display(df_seleccion.head(10))
````

## 5. Evaluación de Impacto de Negocio y Conclusiones para la Gobernanza Analítica

Comparamos el rendimiento predictivo evaluado en el conjunto de prueba (reconvirtiendo las predicciones a dólares reales con `expm1`) para verificar el balance entre **precisión financiera** (¿Qué tan bien predice?) y **parsimonia del modelo** (¿Cuánta complejidad necesitamos? No queremos utilizar 82 variables si podemos obtener los mismos resultados con menos de 10 variables) .

````python
from sklearn.metrics import r2_score, root_mean_squared_error

def auditar_estrategia(X_tr, X_te, nombre):
    modelo = RandomForestRegressor(n_estimators=150, random_state=42)
    modelo.fit(X_tr, y_train_log)
    # Predicción logarítmica revertida a dólares
    pred_dolares = np.expm1(modelo.predict(X_te))
    r2 = r2_score(y_test_orig, pred_dolares)
    rmse = root_mean_squared_error(y_test_orig, pred_dolares)
    return {
        'Estrategia de Modelado': nombre,
        'Variables Utilizadas': X_tr.shape[1],
        'R2 Score (Varianza Explicada)': f"{r2:.2%}",
        'Margen de Error Medio (RMSE)': f"${rmse:,.2f}"
    }

# Comparación
res_completo = auditar_estrategia(X_train, X_test, 'Todas las Variables (30+ vars)')
res_rfe = auditar_estrategia(rfe_selector.transform(X_train), rfe_selector.transform(X_test), 'RFE Compacto (Top 10 vars)')
res_f = auditar_estrategia(selector_f.transform(X_train), selector_f.transform(X_test), 'SelectKBest F-Score (Top 10 vars)')

df_auditoria = pd.DataFrame([res_completo, res_rfe, res_f])
print("Auditoría de rendimiento analítico")
display.display(df_auditoria)
````

### Conclusiones Estratégicas

1. **El valor del Análisis Confirmatorio (CDA):** Demostró formalmente mediante prueba $t$ de Welch y ANOVA que factores como climatización central `CentralAir` y calidad de materiales `OverallQual` no son meras correlaciones muestrales, sino activos con retorno cuantificable sobre el valor inmobiliario.
2. **La potencia de la Ingeniería de Características:** La transformación logarítmica sobre `SalePrice` y la formulación de `TotalLivingSF` y `Age` simplificaron la complejidad de los algoritmos y mejoraron la estabilidad de las estimaciones en producción.
3. **Eficiencia Operacional con RFE:** Reducir de más de 30 variables a únicamente 10 características preserva más del 98% del poder predictivo del negocio, optimizando sustancialmente los costos de nube, almacenamiento y auditoría de modelos.
