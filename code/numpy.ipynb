## Ejemplos motivadores


```python
distances = [10, 15, 17, 26, 20]
times = [0.3, 0.47, 0.55, 1.20, 1.0]
```

```python
# Calcular las velocidades con Python
speeds = []

for i in range(len(distances)):
    speeds.append(distances[i] / times[i])

speeds
```

```python
# El mismo ciclo pero con comprensión de listas
[d / t for d, t in zip(distances, times)]
```

```python
import numpy as np 
```

```python
# Calcular las velocidades con Python con numpy

speeds = np.array(distances) / np.array(times)
speeds
```

```python
# Calcular el precio total de los productos
product_quantities = [13, 5, 6, 10, 11]
prices = [1.2, 6.5, 1.0, 4.8, 5.0]
total = sum([q * p for q, p in zip(product_quantities, prices)])
total
```

```python
# Calcular el precio total de los productos con numpy
total = (np.array(product_quantities) * np.array(prices)).sum()
total
```

# Matrices NumPy: creación, métodos y atributos


## ¿Qué es una matriz NumPy?

El objeto principal de NumPy es la matriz multidimensional homogénea. Es una tabla de elementos (normalmente números), todos del mismo tipo, indexados por una tupla de números enteros positivos. En NumPy las dimensiones se llaman ejes. El número de ejes es el _rank_ o _dimension_.

## Manipulaciones comunes: indexar, cortar y remodelar


### Indexación: obtener y establecer el valor de elementos de matriz individuales


```python
# Matriz de una dimensión (arreglo) de 12 posiciones desde -0.5 hasta 0.6
one_dim = np.linspace(-0.5, 0.6, 12)
one_dim
```

```python
# Quinta posición contando desde la posición-0
one_dim[5]
```

```python
# Reemplaza el valor de la primera posición (posición-0)
one_dim[0] = 1
one_dim
```

```python
# Matriz de dos dimensiones
two_dim = np.array([[3, 5, 2, 4], [7, 6, 5, 5], [1, 6, -1, -1]])
two_dim
```

```python
# Fila 0, columna 3
two_dim[0, 3]
```

```python
# Cambia el valor de la celda de posición: fila-0, columna-0
two_dim[0, 0] = -1
two_dim
```

### Cortar (slicing): obtener y configurar subarreglos más pequeños dentro de un arreglo más grande


```python
one_dim
```

```python
# Corte el arreglo desde la posición 2 hasta la 5 (la última -5- no se incluye)
print(one_dim[2:5])
# Corte el arreglo desde el inicio hasta la posición 5 (la última -5- no se incluye)
print(one_dim[:5])
# Corte el arreglo desde la última posición hasta la posición 5 (de derecha a izquierda)
print(one_dim[-5:])
```

```python
two_dim
```

```python
# Corta desde las filas 0 hasta la 2 (la 2 no se incluye) y desde la columna 0 hasta la 2 (la 2 no se incluye)
two_dim[:2, :2]
```

```python
# Corta incluyendo todas las filas, pero desde la columna 1 hasta la 3 (la 3 no se incluye)
two_dim[:, 1:3]
```

### Remodelación (reshaping) de matrices: cambiar la forma de una matriz determinada


```python
one_dim
```

```python
# Remodela la matriz de una dimensión a otra de dos dimensiones de 2 filas por 6 columnas
one_dim.reshape(2, 6)
```

# Ejemplo: realizar una simulación


## Lanzamientos de monedas


```python
# Lanzar una moneda 1 vez: 0 es cruz, 1 cara.
np.random.randint(low=0, high=2, size=1)
```

```python
# Lanzar una moneda 10 veces: 0 es cruz, 1 cara.
experiment = np.random.randint(0, 2, size=10)
print(experiment)
print(experiment.sum())
```

```python
# Cada columna de esta matriz será una simulación de 10 lanzamientos.
coin_matrix = np.random.randint(0, 2, size=(10000, 10))     # Distribución normal (gaussiana)
# Muestra las primeras 5 filas
coin_matrix[:5, :]
```

```python
# Cuenta las caras por cada fila
counts = coin_matrix.sum(axis=1)
# Muestra los primeros 25 conteos de caras
print(counts[:25])
# Muestra el promedio de caras
print(counts.mean())
# Muestra la mediana de caras
print(np.median(counts))
# Muestra el mínimo y máximo conteo de caras 
print(counts.min(), counts.max())
# Muestra la desviación estándar del conteo de caras
print(counts.std())
```

```python
# Conteo de la frecuencia de caras desde 0 hasta 10
np.bincount(counts)
```

```python
# Define un arreglo vacío para los números que se observaron en el vector de conteo.
unique_numbers = np.arange(0, 11)
unique_numbers
```

```python
observed_times = np.bincount(counts) 
observed_times
```

```python
np.savetxt("coins.csv", coin_matrix, delimiter=",")
```

```python
for n, count in zip(unique_numbers, observed_times):
    print(f"{n} caras observadas en {count} veces ({100 * count / 10000:0.1f}%)")
```

```python
# Biblioteca de visualización
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Graficar el histograma de lanzamientos de monedas que fueron "cara"
plt.hist(counts, bins=np.arange(12) - 0.5, rwidth=0.8, color='steelblue', edgecolor='black')
plt.xticks(range(11))
plt.title("Frecuencia de lanzamientos de monedas que sean cara")
plt.xlabel('Número de Caras')
plt.ylabel('Frecuencia')
plt.show()
```

## Simulación de la rentabilidad de las acciones


```python
# Arreglo de 250 elementos con una media de 0.001 y una distribución estándar de 0.02
# En rendimientos diarios, simulamos 250 días de rendimientos con rendimiento promedio diario de 0.001 = 0.1 %, y una volatilidad diaria de 0.02 = 2 %
returns = np.random.normal(0.001, 0.02, 250)
```

```python
# Primeros 20 días
returns[:20]
```

```python
# Precio inicial
initial_price = 100
```

```python
# Convertir los rendimientos simulados en una trayectoria de precios, cumsum() significa suma acumulada,
#  np.exp(returns.cumsum()), convierte los rendimientos acumulados en factores de crecimiento.
# Esto es una forma sencilla de hacer una simulación de Monte Carlo de precios.
price = initial_price * np.exp(returns.cumsum())
```

```python
# Primeros 20 rendimientos acumulados
price[:20]
```

```python
# Grafica el rendimiento acumulado
plt.plot(price)
plt.grid()
plt.title('Rendimiento Acumulado Simulación 1')
plt.xlabel('Días')
plt.ylabel('Rendimiento')
plt.show()
```

```python
# Segunda simulación
returns = np.random.normal(0.001, 0.02, 250)
initial_price = 100
price = initial_price * np.exp(returns.cumsum())
plt.plot(price)
plt.grid()
plt.title('Rendimiento Acumulado Simulación 2')
plt.xlabel('Días')
plt.ylabel('Rendimiento')
plt.show()
```
