## La serie de Pandas

La estructura de datos **Series** en Pandas es un <i>array unidimensional etiquetado</i>.

* Los datos en el array pueden ser de cualquier tipo (enteros, cadenas, números de coma flotante, objetos de Python, etc.).
* Los datos dentro del array son homogéneos.
* Los objetos Pandas Series siempre tienen un índice: esto les confiere propiedades similares a las de un ndarray y un diccionario.
    
<img src="img/series3.png">

## Creando una serie de pandas

Hay muchas formas de crear objetos de la serie Pandas, algunas de las más comunes son:
- Creación a partir de una lista.
- Creación a partir de un diccionario.
- Creación a partir de un ndarray.
- De una fuente externa como un archivo

```python
# Define los datos y el índice como listas
temperature = [18, 7, 6, 20, 8, -1, 3]
days = ['Lun','Mar','Mie','Jue','Vie','Sab','Dom']

# Crea la serie
series_from_list = pd.Series(temperature, index=days)
series_from_list
```

```python
# Desde un diccionario
my_dict = {'Lun': 18, 'Mar': 7, 'Mie': 6, 'Jue': 20, 'Vie': 8, 'Sab': -1, 'Dom': 3}
series_from_dict = pd.Series(my_dict)
series_from_dict
```

```python
# Desde un arreglo NumPy
my_array = np.linspace(0, 10, 15)
series_from_ndarray = pd.Series(my_array)
series_from_ndarray
```

```python
# Las operaciones vectorizadas también funcionan en las series de pandas.
np.exp(series_from_list)
```

### Compara a los empleados con baja y muy alta satisfacción laboral según las siguientes variables: edad, departamento, distancia al domicilio, salario por hora, ingresos mensuales y años en la empresa.

**Operaciones de agrupación**: Operación Split-Apply-Combine (Dividir-Aplicar-Combinar).

Por operaciones **agrupación** o **group by** nos referimos a un proceso que incluye uno o más de los siguientes pasos:

- **Splitting (Dividir)** los datos en grupos según algún criterio.
- **Applying (Aplicar)** una función a cada grupo de forma independiente.
- **Combining (Combinar)** los resultados en una estructura de datos.

<b>Paso 1 (División):</b> La operación <i>groupby</i> <b><i>divide</b></i> el dataframe en un grupo de dataframes según ciertos criterios. Tenga en cuenta que el objeto agrupado <i>no</i> es un dataframe. Es un objeto GroupBy. Tiene una estructura similar a un diccionario y también es iterable.

<img src="img/groupby1.jpg">

<b>Paso 2 (Analizar):</b> Una vez que tenemos un objeto agrupado, podemos <b><i>aplicar</b></i> funciones o ejecutar análisis a cada grupo, conjunto de grupos o al grupo completo.

<img src="img/groupby2.jpg">

<b>Paso 3 (Combinar):</b> También podemos <b><i>combinar</b></i> los resultados del análisis en una o varias estructuras de datos nuevas.

<img src="img/groupby3.jpg">

Dado que solo nos interesan los empleados con niveles de satisfacción laboral "bajos" y "muy altos", vamos a crear un nuevo DataFrame que contenga únicamente esas observaciones.

```python
subset_of_interest = data.loc[(data['JobSatisfaction'] == "Bajo") | (data['JobSatisfaction'] == "Muy Alto")]
subset_of_interest.shape
```

Dado que nuestra variable JobSatisfaction tiene 4 categorías, estas categorías se han mantenido en la serie de este nuevo DataFrame:

```python
subset_of_interest['JobSatisfaction'].value_counts()
```

Eliminemos las categorías que no vamos a utilizar:

```python
subset_of_interest['JobSatisfaction'] = (
    subset_of_interest['JobSatisfaction']
    .cat.remove_unused_categories()
)
```

```python
subset_of_interest['JobSatisfaction'].value_counts()
```

Ahora que solo tenemos a los empleados que nos interesan, podemos compararlos según las variables que queríamos. Primero, dividamos nuestro nuevo DataFrame en grupos.

```python
grouped = subset_of_interest.groupby('JobSatisfaction', observed=True)
```

### Comparación de las medias en todas las variables numéricas

Aunque solo solicitamos algunas columnas específicas, para que el director de recursos humanos tenga una mejor idea de cómo se comparan estos grupos en función de diferentes variables, vamos a crear un DataFrame que contenga la media de cada variable numérica de nuestro conjunto de datos.       

```python
# Obtención de las columnas numéricas
numeric_cols = subset_of_interest.select_dtypes(include=[np.number]).columns
```

```python
# Crear un DataFrame vacío
mean_comparison_df = pd.DataFrame(columns=numeric_cols, index=['Bajo', 'Muy Alto'])
mean_comparison_df
```

```python
grouped['Age'].mean()
```

```python
# Rellenar el DataFrame
for var in numeric_cols:
    mean_comparison_df[var] = grouped[var].mean()
```

```python
mean_comparison_df
```

```python
mean_comparison_df = mean_comparison_df.transpose()
mean_comparison_df
```

### Hagamos una visualización

```python
mean_comparison_df.plot(kind='bar', figsize=(13,4), title="Comparación de Medias")
```
